```python

#functions-2 

def get_pay(num_hours):
    #pre-tax pay , based on receiving $15/hour

    pay_pretax = num_hours * 15

    pay_aftertax = pay_pretax * .88
    return pay_aftertax
```

```python
#Now calculate on the basis on working 40 hours 

pay_fulltime = get_pay(40)
print(pay_fulltime)
```

```python
pay_parttime = get_pay(30)
print(pay_parttime)
```

```python
#Functions with multiple arguments

def get_pay_with_more_inputs(Num_hours, hourly_wage, tax_bracket):
    pay_pretax = Num_hours * hourly_wage

    pay_aftertax = pay_pretax * (1 - tax_bracket)
    return pay_aftertax
```

```python
High_pay_aftertax = get_pay_with_more_inputs(34, 30, .23)
print(High_pay_aftertax)
```

```python
good_pay_aftertax = get_pay_with_more_inputs(59, 20, .12)
print(good_pay_aftertax)
```
